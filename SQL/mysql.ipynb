### 14. WHERE vs HAVING

This is very important for interviews.

#### WHERE

Filters **rows before grouping**.

```sql
SELECT *
FROM employees
WHERE salary > 50000;
```

---

#### HAVING

Filters **groups after GROUP BY**.

```sql
SELECT
    department,
    AVG(salary) AS avg_salary
FROM employees
GROUP BY department
HAVING AVG(salary) > 50000;
```

#### Remember

```text
WHERE  → Filter rows
GROUP BY → Create groups
HAVING → Filter groups
```

### 17. CASE WHEN vs np.where()

Used for conditional logic.

#### SQL

```sql
SELECT
    name,
    salary,
    CASE
        WHEN salary >= 60000 THEN 'High'
        WHEN salary >= 40000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_category
FROM employees;
```

#### Pandas

```python
import numpy as np

df["salary_category"] = np.where(
    df["salary"] >= 60000,
    "High",
    "Low"
)
```

For multiple conditions:

```python
conditions = [
    df["salary"] >= 60000,
    df["salary"] >= 40000
]

choices = [
    "High",
    "Medium"
]

df["salary_category"] = np.select(
    conditions,
    choices,
    default="Low"
)
```

### 20. SQL CASE vs Pandas apply()

For more complex row-wise logic.

#### SQL

```sql
SELECT
    name,
    CASE
        WHEN age < 18 THEN 'Minor'
        WHEN age < 60 THEN 'Adult'
        ELSE 'Senior'
    END AS age_group
FROM customers;
```

#### Pandas

```python
def age_group(age):
    if age < 18:
        return "Minor"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"

df["age_group"] = df["age"].apply(age_group)
```

### 23. CTE

CTE = Common Table Expression.

Used to make complex SQL queries easier to read.

#### SQL

```sql
WITH dept_salary AS (
    SELECT
        department,
        AVG(salary) AS avg_salary
    FROM employees
    GROUP BY department
)

SELECT *
FROM dept_salary
WHERE avg_salary > 50000;
```

Pandas usually performs the equivalent work through intermediate DataFrames:

```python
dept_salary = (
    df.groupby("department")["salary"]
      .mean()
      .reset_index(name="avg_salary")
)

result = dept_salary[
    dept_salary["avg_salary"] > 50000
]
```

### 24. Window Functions

Window functions perform calculations across related rows without collapsing the rows.

Important functions:

```text
ROW_NUMBER()
RANK()
DENSE_RANK()
LAG()
LEAD()
SUM() OVER()
AVG() OVER()
```

---

#### Example: Rank Employees by Salary

### SQL

```sql
SELECT
    name,
    department,
    salary,
    RANK() OVER(
        PARTITION BY department
        ORDER BY salary DESC
    ) AS salary_rank
FROM employees;
```

### Pandas

```python
df["salary_rank"] = (
    df.groupby("department")["salary"]
      .rank(method="min", ascending=False)
)
```

#### When to use?

Useful for:

- Top N analysis
- Ranking products
- Ranking employees
- Previous/next period analysis
- Running totals

### 17. CASE WHEN vs np.where()

For multiple conditions:

### 23. CTE

Used to make complex SQL queries easier to read.

#### SQL

```sql
WITH dept_salary AS (
    SELECT
        department,
        AVG(salary) AS avg_salary
    FROM employees
    GROUP BY department
)

SELECT *
FROM dept_salary
WHERE avg_salary > 50000;
```

# 1. Primary Key

A **Primary Key** uniquely identifies each row in a table.

```sql
CREATE TABLE customers (
    customer_id INT PRIMARY KEY,
    name VARCHAR(50),
    city VARCHAR(50)
);
```

```text
customer_id → Primary Key
```

### Important Properties

- Must be unique
- Cannot contain NULL
- A table generally has one primary key

SQL:

```sql
CREATE TABLE orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    amount DECIMAL(10,2),

    FOREIGN KEY (customer_id)
    REFERENCES customers(customer_id)
);
```

### Relationship

```text
customers
    |
    | customer_id
    ↓
orders
```

### When to use?

When tables have relationships.

```text
Customer → Orders
Product → Sales
Employee → Department
Student → Courses
```

# 39. Query Execution Order

This is very important for understanding SQL.

Although we write:

```sql
SELECT
FROM
WHERE
GROUP BY
HAVING
ORDER BY
LIMIT
```

SQL logically processes it approximately as:

```text
FROM
  ↓
WHERE
  ↓
GROUP BY
  ↓
HAVING
  ↓
SELECT
  ↓
ORDER BY
  ↓
LIMIT
```

### Example

```sql
SELECT
    city,
    AVG(salary) AS avg_salary
FROM employees
WHERE age > 25
GROUP BY city
HAVING AVG(salary) > 50000
ORDER BY avg_salary DESC
LIMIT 5;
```

# 44. SQL for Business Questions

A Data Analyst should learn to convert business questions into SQL queries.

### Question 1

> What is the total revenue?

```sql
SELECT SUM(revenue)
FROM sales;
```

---

### Question 2

> Which product generated the highest revenue?

```sql
SELECT
    product,
    SUM(revenue) AS total_revenue
FROM sales
GROUP BY product
ORDER BY total_revenue DESC
LIMIT 1;
```

---

### Question 3

> How many unique customers placed orders?

```sql
SELECT COUNT(DISTINCT customer_id)
FROM orders;
```

---

### Question 4

> Which month had the highest sales?

```sql
SELECT
    MONTH(order_date) AS month,
    SUM(sales) AS total_sales
FROM orders
GROUP BY MONTH(order_date)
ORDER BY total_sales DESC
LIMIT 1;
```

---

### Question 5

> Find customers who never placed an order.

```sql
SELECT
    c.customer_id,
    c.name
FROM customers c
LEFT JOIN orders o
    ON c.customer_id = o.customer_id
WHERE o.customer_id IS NULL;
```
