### 14. WHERE vs HAVING

This is very important for interviews.

#### WHERE

Filters **rows before grouping**.

```sql
SELECT *
FROM employees
WHERE salary > 50000;
```

---

#### HAVING

Filters **groups after GROUP BY**.

```sql
SELECT
    department,
    AVG(salary) AS avg_salary
FROM employees
GROUP BY department
HAVING AVG(salary) > 50000;
```

#### Remember

```text
WHERE  → Filter rows
GROUP BY → Create groups
HAVING → Filter groups
```

### 17. CASE WHEN vs np.where()

Used for conditional logic.

#### SQL

```sql
SELECT
    name,
    salary,
    CASE
        WHEN salary >= 60000 THEN 'High'
        WHEN salary >= 40000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_category
FROM employees;
```

#### Pandas

```python
import numpy as np

df["salary_category"] = np.where(
    df["salary"] >= 60000,
    "High",
    "Low"
)
```

For multiple conditions:

```python
conditions = [
    df["salary"] >= 60000,
    df["salary"] >= 40000
]

choices = [
    "High",
    "Medium"
]

df["salary_category"] = np.select(
    conditions,
    choices,
    default="Low"
)
```

### 20. SQL CASE vs Pandas apply()

For more complex row-wise logic.

#### SQL

```sql
SELECT
    name,
    CASE
        WHEN age < 18 THEN 'Minor'
        WHEN age < 60 THEN 'Adult'
        ELSE 'Senior'
    END AS age_group
FROM customers;
```

#### Pandas

```python
def age_group(age):
    if age < 18:
        return "Minor"
    elif age < 60:
        return "Adult"
    else:
        return "Senior"

df["age_group"] = df["age"].apply(age_group)
```

### 23. CTE

CTE = Common Table Expression.

Used to make complex SQL queries easier to read.

#### SQL

```sql
WITH dept_salary AS (
    SELECT
        department,
        AVG(salary) AS avg_salary
    FROM employees
    GROUP BY department
)

SELECT *
FROM dept_salary
WHERE avg_salary > 50000;
```

Pandas usually performs the equivalent work through intermediate DataFrames:

```python
dept_salary = (
    df.groupby("department")["salary"]
      .mean()
      .reset_index(name="avg_salary")
)

result = dept_salary[
    dept_salary["avg_salary"] > 50000
]
```

### 24. Window Functions

Window functions perform calculations across related rows without collapsing the rows.

Important functions:

```text
ROW_NUMBER()
RANK()
DENSE_RANK()
LAG()
LEAD()
SUM() OVER()
AVG() OVER()
```

---

#### Example: Rank Employees by Salary

### SQL

```sql
SELECT
    name,
    department,
    salary,
    RANK() OVER(
        PARTITION BY department
        ORDER BY salary DESC
    ) AS salary_rank
FROM employees;
```

### Pandas

```python
df["salary_rank"] = (
    df.groupby("department")["salary"]
      .rank(method="min", ascending=False)
)
```

#### When to use?

Useful for:

- Top N analysis
- Ranking products
- Ranking employees
- Previous/next period analysis
- Running totals

Use LEFT JOIN when you want **all records from the main table**, even if matching data does not exist in the second table.

Example:

```text
All customers
+
Their orders if available
```

### 17. CASE WHEN vs np.where()

Used for conditional logic.

#### SQL

```sql
SELECT
    name,
    salary,
    CASE
        WHEN salary >= 60000 THEN 'High'
        WHEN salary >= 40000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_category
FROM employees;
```

#### Pandas

```python
import numpy as np

df["salary_category"] = np.where(
    df["salary"] >= 60000,
    "High",
    "Low"
)
```

For multiple conditions:

```python
conditions = [
    df["salary"] >= 60000,
    df["salary"] >= 40000
]

choices = [
    "High",
    "Medium"
]

df["salary_category"] = np.select(
    conditions,
    choices,
    default="Low"
)
```

### 23. CTE

CTE = Common Table Expression.

Used to make complex SQL queries easier to read.

#### SQL

```sql
WITH dept_salary AS (
    SELECT
        department,
        AVG(salary) AS avg_salary
    FROM employees
    GROUP BY department
)

SELECT *
FROM dept_salary
WHERE avg_salary > 50000;
```

Pandas usually performs the equivalent work through intermediate DataFrames:

```python
dept_salary = (
    df.groupby("department")["salary"]
      .mean()
      .reset_index(name="avg_salary")
)

result = dept_salary[
    dept_salary["avg_salary"] > 50000
]
```
