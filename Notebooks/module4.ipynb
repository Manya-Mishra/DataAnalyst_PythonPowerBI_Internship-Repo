## 📊 Module 4: Excel for Data Analysis

## 1. What is Excel?

**Microsoft Excel** is a spreadsheet application used to:

* Store data
* Clean data
* Organize data
* Perform calculations
* Analyze data
* Create charts
* Build reports
* Create PivotTables
* Transform data using Power Query

For a Data Analyst, Excel is especially useful for **small-to-medium datasets, quick analysis, reporting, business calculations, and exploratory work**.

## 4. Cell References

Every Excel cell has an address.

Example:

```text
A1
B2
C5
D10
```

Where:

```text
A → Column
1 → Row
```

Example:

```text
A1 = Customer Name
B1 = Sales
```

---

## 5. Range

A range represents multiple cells.

Example:

```text
A1:A10
```

means:

```text
A1
A2
A3
...
A10
```

Another example:

```text
A1:C10
```

represents a rectangular range from A1 to C10.

---

## 6. Relative Cell Reference

A relative reference changes when a formula is copied.

Example:

```excel
=A2*B2
```

If copied to the next row:

```excel
=A3*B3
```

The references automatically change.

---

## 7. Absolute Cell Reference

An absolute reference remains fixed when copied.

Use:

```text
$
```

Example:

```excel
=A2*$B$1
```

Here:

```text
A2 → Changes
$B$1 → Remains fixed
```

### When to use?

Useful when one cell contains a fixed value such as:

```text
Tax Rate
Discount Rate
Commission Rate
Exchange Rate
```

Example:

| Product | Price | Tax Rate |
| ------- | ----: | -------: |
| A       |  1000 |      18% |
| B       |  2000 |      18% |

Formula:

```excel
=B2*$C$2
```

---

## 8. Mixed Cell Reference

A mixed reference fixes either the row or column.

Examples:

```excel
$A1
A$1
```

### `$A1`

Column A is fixed, row can change.

### `A$1`

Row 1 is fixed, column can change.

# 17. COUNTIF()

Counts cells that satisfy one condition.

### Syntax

```excel
=COUNTIF(range, criteria)
```

Example:

```excel
=COUNTIF(C2:C100,"Delhi")
```

Counts customers from Delhi.

Another example:

```excel
=COUNTIF(D2:D100,">50000")
```

Counts salaries greater than ₹50,000.

---

# 18. COUNTIFS()

Counts records satisfying multiple conditions.

### Syntax

```excel
=COUNTIFS(range1, criteria1, range2, criteria2)
```

Example:

```excel
=COUNTIFS(B2:B100,"Delhi",C2:C100,">50000")
```

Meaning:

```text
City = Delhi
AND
Salary > 50000
```

Meaning:

> Calculate total sales for Pizza.

# 23. IF Function

`IF()` performs conditional logic.

### Syntax

```excel
=IF(condition, value_if_true, value_if_false)
```

Example:

```excel
=IF(B2>=50000,"High","Low")
```

Meaning:

```text
Salary >= 50,000
      ↓
   High

Otherwise
      ↓
   Low
```

---

# 24. Nested IF

Multiple conditions can be checked.

```excel
=IF(B2>=80000,"High",
    IF(B2>=50000,"Medium","Low"))
```

Result:

```text
>= 80000 → High
>= 50000 → Medium
< 50000  → Low
```

For many conditions, `IFS()` can often be easier to read.

# 53. Excel Lookup Functions

Lookup functions are extremely important for Data Analysts.

Common lookup functions:

```text
VLOOKUP
XLOOKUP
HLOOKUP
INDEX
MATCH
```

Microsoft describes XLOOKUP as a lookup function that can match a value and return a related value, including cases where the lookup column is not the first column.

---

# 54. VLOOKUP()

VLOOKUP searches for a value in the **first column of a table** and returns a value from another column.

### Syntax

```excel
=VLOOKUP(
    lookup_value,
    table_array,
    col_index_num,
    FALSE
)
```

Example:

| Product_ID | Product | Price |
| ---------- | ------- | ----: |
| P101       | Pizza   |   300 |
| P102       | Burger  |   200 |
| P103       | Pasta   |   250 |

Formula:

```excel
=VLOOKUP(A2,$F$2:$H$4,2,FALSE)
```

If:

```text
A2 = P102
```

Result:

```text
Burger
```

---

# 55. VLOOKUP — Important Points

```text
V → Vertical
LOOKUP → Search
```

`FALSE` means:

```text
Exact Match
```

`TRUE` means:

```text
Approximate Match
```

For most beginner data-analysis lookup tasks, exact matching is safer when IDs should match exactly:

```excel
FALSE
```

---

# 56. Limitations of VLOOKUP

Traditional VLOOKUP has limitations:

* Lookup column generally needs to be the first column of the selected table
* Column numbers are manually specified
* Inserting columns can make formulas harder to maintain
* It is less flexible than XLOOKUP

---

# 57. XLOOKUP()

XLOOKUP is a more flexible lookup function available in modern Excel versions.

### Syntax

```excel
=XLOOKUP(
    lookup_value,
    lookup_array,
    return_array,
    [if_not_found]
)
```

Example:

```excel
=XLOOKUP(
    A2,
    F2:F4,
    G2:G4,
    "Not Found"
)
```

Meaning:

```text
Search A2 in F2:F4
        ↓
Return matching value from G2:G4
```

---

# 58. XLOOKUP Example

Product table:

| Product_ID | Product | Price |
| ---------- | ------- | ----: |
| P101       | Pizza   |   300 |
| P102       | Burger  |   200 |
| P103       | Pasta   |   250 |

If:

```text
A2 = P102
```

Formula:

```excel
=XLOOKUP(A2,F2:F4,G2:G4,"Not Found")
```

Result:

```text
Burger
```

---

# 59. VLOOKUP vs XLOOKUP

| Feature                | VLOOKUP                 | XLOOKUP                  |
| ---------------------- | ----------------------- | ------------------------ |
| Exact match            | Yes                     | Yes                      |
| Flexible lookup        | Limited                 | Better                   |
| Lookup direction       | Primarily left-to-right | More flexible            |
| Error handling         | Usually IFERROR         | Built-in argument        |
| Column number required | Yes                     | No                       |
| Modern recommendation  | Legacy/common           | Preferred when available |

# 123. Common Interview Questions

### Q1. What is the difference between COUNT and COUNTA?

```text
COUNT  → Counts numeric cells
COUNTA → Counts non-empty cells
```

---

### Q2. What is the difference between COUNTIF and COUNTIFS?

```text
COUNTIF  → One condition
COUNTIFS → Multiple conditions
```

---

### Q3. What is the difference between SUMIF and SUMIFS?

```text
SUMIF  → One condition
SUMIFS → Multiple conditions
```

---

### Q4. What is VLOOKUP?

VLOOKUP searches for a value in the first column of a selected table and returns a corresponding value from another column.

---

### Q5. Why is XLOOKUP useful?

It provides more flexible lookup behavior than traditional VLOOKUP and allows separate lookup and return arrays, along with an optional not-found result.

---

### Q6. What is a PivotTable?

A PivotTable is an Excel feature used to quickly summarize and analyze data by grouping fields and applying aggregations such as sum, count, or average.

---

### Q7. What is Power Query?

Power Query is an Excel data-import and transformation tool used to connect to data sources, clean and transform data, combine datasets, and load the result for analysis.

---

### Q8. What is the difference between Merge and Append in Power Query?

```text
Merge  → Combines tables using matching columns/keys
Append → Stacks rows from tables
```

---

### Q9. What is the difference between Power Query and Power Pivot?

```text
Power Query → Import + Clean + Transform

Power Pivot → Data Model + Relationships + DAX
```

Microsoft describes these as complementary tools: Power Query shapes/imports data, while Power Pivot adds modeling capabilities.

---
