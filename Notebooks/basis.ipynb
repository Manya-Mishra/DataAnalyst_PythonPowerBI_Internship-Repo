## What is Data Analysis?

### Data Analysis is the process of collecting, cleaning, transforming, and interpreting data to discover useful insights, identify patterns, and support decision-making.<br> In simple words, Data analysis means turning raw data into meaningful insights.

Example

A retail company has sales data for the last year.

A Data Analyst can:

Identify the best-selling products
Find peak sales months
Analyze customer purchasing behavior
Recommend strategies to increase revenue

## Data-Driven Decision Making
Definition

Data-driven decision making (DDDM) refers to making decisions based on analyzed data rather than intuition or guesswork.

Traditional Approach

"Sales seem low this month."

Data-Driven Approach

"Sales decreased by 12% because customer purchases dropped in the North region."

Advantages
* Higher accuracy
* Reduced uncertainty
* Better business outcomes
* Objective decision-making

Example 1: Restaurant

❌ Guess-Based Decision
"We think customers like Pizza."

✅ Data-Driven Decision
"Pizza contributes 35% of total revenue and has the highest repeat orders."

Example 2: Online Learning Platform

❌ Guess-Based Decision
"Students probably like video lectures."

✅ Data-Driven Decision
"Students spend 70% more time on video lessons than text lessons."

## Data Analysis Lifecycle

The Data Analysis Lifecycle describes the step-by-step process followed by analysts to solve business problems using data.

### Step 1: Business Understanding

Purpose
<br>Understand the problem before analyzing data.

Questions to Ask
* What is the business problem?
* What are the goals?
* What decisions need to be made?

Example 1 - 
<br>A telecom company wants to reduce customer churn.

Problem:
<br><b>"Why are customers leaving the service?"

Example 2 -
<br>Gym Membership

Problem:
<br><b>Many members stop coming after 3 months.</b>

Business Question:
<br><b>Why are members leaving?</b>

Goal:
<br><b>Increase member retention.</b>

### Step 2: Data Collection
Purpose
<br>Gather relevant data from different sources.

Sources
* Databases
* Excel files
* APIs
* Surveys
* Websites
* IoT devices

Example -
<br>Telecom Company

<b>

</b>

### Step 3: Data Cleaning
Purpose

Prepare data for analysis.

Common Issues
* Missing values
* Duplicate records
* Incorrect data types
* Inconsistent formats
* Outliers

|------------|----:|
| 101 | 25 |
| 102 | Null |

<b>After Cleaning</b>

| Customer_ID | Age |
|------------|----:|
| 101 | 25 |
| 102 | 24 |

<br>
Example - <br>

<b>Raw Data</b>

|------------|----:|
