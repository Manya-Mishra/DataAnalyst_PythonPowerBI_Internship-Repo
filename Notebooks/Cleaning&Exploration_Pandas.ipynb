# Data Cleaning & Data Exploration in Pandas

Data Cleaning and Data Exploration are two important stages of the Data Analysis process.

- **Data Cleaning:** Improves data quality by handling missing values, duplicates, incorrect data types, inconsistent formatting, and outliers.
- **Data Exploration (EDA):** Helps understand the dataset by summarizing, visualizing, and identifying patterns or relationships.

---

# Sample Dataset

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Alice"],
    "Age": [25, None, 30, 28, 25],
    "Salary": [50000, 60000, None, 55000, 50000],
    "Department": ["HR", "IT", "Finance", "IT", "HR"]
}

df = pd.DataFrame(data)

print(df)
```

Output

| Name | Age | Salary | Department |
|------|-----|--------|------------|
| Alice | 25 | 50000 | HR |
| Bob | NaN | 60000 | IT |
| Charlie | 30 | NaN | Finance |
| David | 28 | 55000 | IT |
| Alice | 25 | 50000 | HR |

# Additional Pandas Functions for Data Cleaning & Data Exploration

# Data Cleaning

## 1. copy()

Creates a copy of the DataFrame.

### Syntax

```python
df_copy = df.copy()
```

### Why?

Prevents accidental modification of the original dataset.

---

## 2. copy(deep=True)

Creates an independent copy of the data.

```python
df_copy = df.copy(deep=True)
```

---

## 3. drop()

Removes rows or columns.

### Remove Column

```python
df.drop("Salary", axis=1)
```

### Remove Row

```python
df.drop(0)
```

---

## 4. insert()

Adds a new column at a specified position.

```python
df.insert(1, "Gender", ["F","M","M","M"])
```

---

## 5. assign()

Creates a new column.

```python
df = df.assign(Bonus=df["Salary"]*0.10)
```

---

## 6. apply()

Applies a function to rows or columns.

```python
df["Salary"] = df["Salary"].apply(lambda x: x*1.1)
```

---

## 7. map()

Maps values in a Series.

```python
df["Department"] = df["Department"].map({
    "HR":"Human Resource",
    "IT":"Information Technology"
})
```

---

## 8. applymap()

Applies a function to every element in the DataFrame.

```python
df = df.applymap(str)
```

> **Note:** In newer versions of pandas, `DataFrame.map()` is preferred over `applymap()` for element-wise operations.

---

## 9. strip()

Removes leading/trailing spaces.

```python
df["Name"] = df["Name"].str.strip()
```

---

## 10. lower()

Converts text to lowercase.

```python
df["Department"] = df["Department"].str.lower()
```

---

## 11. upper()

Converts text to uppercase.

```python
df["Department"] = df["Department"].str.upper()
```

---

## 12. title()

Converts to title case.

```python
df["Name"] = df["Name"].str.title()
```

---

## 13. replace()

Replace multiple values.

```python
df.replace({
    "Male":"M",
    "Female":"F"
})
```

---

## 14. clip()

Limits values to a specified range.

```python
df["Salary"] = df["Salary"].clip(lower=30000, upper=100000)
```

Useful for handling outliers.

---

## 15. interpolate()

Fills missing values by interpolation.

```python
df.interpolate()
```

Useful in time-series data.
