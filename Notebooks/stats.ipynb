## 📊 Module 2: Fundamentals of Data

### 1. What is Data?

**Data** is a collection of facts, observations, measurements, or values that can be used to understand something and make decisions.

#### Example

A company stores customer information:

| Customer_ID | Name  | Age | City  | Purchase |
| ----------- | ----- | --: | ----- | -------: |
| 101         | Rahul |  22 | Delhi |      500 |
| 102         | Priya |  25 | Noida |      750 |
| 103         | Aman  |  21 | Delhi |      300 |

Each value in this table is **data**.

A Data Analyst uses this data to answer questions such as:

* Which city has the highest sales?
* What is the average customer age?
* Which customers spent more than ₹500?
* Which product or category generates the most revenue?

---

### 2. Data vs Information vs Insight

These three concepts are important in Data Analysis.

### Data

Raw facts without interpretation.

**Example:**

```text
500, 750, 300, 900, 450
```

### Information

Processed or organized data that gives meaning.

**Example:**

```text
Average purchase = ₹580
```

### Insight

A meaningful conclusion that can support decision-making.

**Example:**

```text
Customers from Noida have a higher average purchase value than customers from Delhi.
```

#### Simple Flow

```text
Raw Data
   ↓
Processing
   ↓
Information
   ↓
Analysis
   ↓
Insight
   ↓
Decision
```

---

### 3. Types of Data

Data can be classified in different ways.

The two most important classifications for Data Analysts are:

1. **Qualitative vs Quantitative**
2. **Categorical vs Numerical**

---

### 4. Qualitative Data

**Qualitative data** describes characteristics, qualities, or categories.

It generally answers:

> **What type?**

#### Examples

```text
Gender
City
Department
Product Category
Customer Feedback
Payment Method
```

```text
City → Qualitative
Payment → Qualitative
```

Qualitative data is usually non-numerical, although categories can sometimes be represented using numbers.

---

### 5. Quantitative Data

**Quantitative data** represents numerical values that can be measured or counted.

It answers:

> **How much?**
>
> **How many?**

#### Examples

```text
Age
Salary
Revenue
Quantity
Temperature
Height
Weight
Number of Orders
```

Example:

| Customer | Age | Salary |
| -------- | --: | -----: |
| A        |  22 |  30000 |
| B        |  25 |  45000 |
| C        |  28 |  50000 |

Here:

```text
Age → Quantitative
Salary → Quantitative
```

### 9. Discrete Data

**Discrete data** consists of countable values.

Usually, it represents things that can be counted.

#### Examples

```text
Number of customers
Number of orders
Number of products
Number of employees
Number of complaints
```

For example:

```text
Number of Orders = 25
```

You can have:

```text
24 orders
25 orders
26 orders
```

But normally not:

```text
25.7 orders
```

#### Key idea

> **Discrete = Countable**

### 10. Continuous Data

**Continuous data** can take any value within a range and is usually obtained through measurement.

#### Examples

```text
Height
Weight
Temperature
Distance
Time
AQI
Revenue
```

For example:

```text
Temperature = 25.6°C
Height = 172.5 cm
Weight = 63.7 kg
```

#### Key idea

> **Continuous = Measurable**

---

### 11. Discrete vs Continuous

| Feature | Discrete              | Continuous           |
| ------- | --------------------- | -------------------- |
| Meaning | Countable values      | Measurable values    |
| Values  | Usually whole numbers | Can contain decimals |
| Example | Number of orders      | Temperature          |
| Example | Number of customers   | Height               |
| Example | Number of products    | Weight               |

#### Easy Trick

```text
COUNT → Discrete
MEASURE → Continuous
```

---

### 12. Nominal Data

**Nominal data** consists of categories that have **no natural order or ranking**.

#### Examples

```text
Gender
City
Department
Blood Group
Payment Method
Product Category
```

For example:

```text
Payment Method:
UPI
Cash
Card
```

There is no meaningful order such as:

```text
UPI > Cash > Card
```

Therefore, it is nominal data.

---

### 13. Ordinal Data

**Ordinal data** consists of categories that have a meaningful order or ranking.

#### Examples

```text
Customer Satisfaction:
Poor
Average
Good
Excellent
```

There is an order:

```text
Poor < Average < Good < Excellent
```

Another example:

```text
Education:
School
Bachelor's
Master's
PhD
```

There is a meaningful progression.

#### Important Point

Although ordinal categories have an order, the **difference between categories is not necessarily equal**.

For example:

```text
Poor → Average
```

does not necessarily represent the same numerical difference as:

```text
Good → Excellent
```

---

### 14. Nominal vs Ordinal

| Feature    | Nominal        | Ordinal            |
| ---------- | -------------- | ------------------ |
| Categories | Yes            | Yes                |
| Order      | No             | Yes                |
| Ranking    | No             | Yes                |
| Example    | City           | Satisfaction       |
| Example    | Gender         | Education Level    |
| Example    | Payment Method | Small/Medium/Large |

---

### 15. Interval Data

**Interval data** is numerical data where:

* Values have a meaningful order
* Differences between values are meaningful
* There is **no true zero**

#### Example

Temperature in Celsius:

```text
10°C
20°C
30°C
```

The difference between:

```text
10°C and 20°C = 10°C
20°C and 30°C = 10°C
```

is meaningful.

However, `0°C` does not mean the complete absence of temperature.

Therefore, Celsius temperature is an example of interval data.

---

### 16. Ratio Data

**Ratio data** has:

* Meaningful order
* Equal intervals
* A true zero

#### Examples

```text
Age
Weight
Height
Salary
Revenue
Distance
Quantity
```

For example:

```text
₹20,000
₹40,000
```

₹40,000 is twice ₹20,000.

Similarly:

```text
10 kg
20 kg
```

20 kg is twice 10 kg.

This is possible because ratio data has a meaningful zero.

---

### 17. Interval vs Ratio

| Feature                  | Interval            | Ratio  |
| ------------------------ | ------------------- | ------ |
| Order                    | Yes                 | Yes    |
| Equal intervals          | Yes                 | Yes    |
| True zero                | No                  | Yes    |
| Example                  | Celsius temperature | Weight |
| Example                  | Fahrenheit          | Salary |
| Can say "twice as much"? | No                  | Yes    |

#### Easy Memory Trick

```text
Interval → No true zero
Ratio → Real/true zero
```

---

### 18. Levels of Measurement

The four major measurement levels are:

```text
Nominal
   ↓
Ordinal
   ↓
Interval
   ↓
Ratio
```

| Level    | Order | Equal Difference | True Zero | Example        |
| -------- | ----- | ---------------- | --------- | -------------- |
| Nominal  | ❌     | ❌                | ❌         | City           |
| Ordinal  | ✅     | ❌                | ❌         | Satisfaction   |
| Interval | ✅     | ✅                | ❌         | Temperature °C |
| Ratio    | ✅     | ✅                | ✅         | Salary         |

# 49. Population vs Sample

## Population

The **population** is the complete group we want to study.

Example:

```text
All customers of a company
```

## Sample

A **sample** is a smaller subset selected from the population.

Example:

```text
1,000 customers selected from 1,00,000 customers
```

### Flow

```text
Population
1,00,000 customers
       ↓
Sample
1,000 customers
       ↓
Analysis
       ↓
Estimate / Conclusion
```

---

* Expensive
* Time-consuming

* Sometimes impossible

```text
1,00,00,000 customers
```

we may survey:

```text
10,000 customers
```

and analyze their responses.

# 52. Types of Sampling

Important sampling methods include:

### 1. Simple Random Sampling

Every member has an equal chance of being selected.

### 2. Stratified Sampling

Population is divided into groups and samples are selected from each group.

Example:

```text
Population
├── Male
├── Female
└── Other
```

Take samples from each group.

### 3. Systematic Sampling

Select every `k`th observation.

Example:

```text
Every 10th customer
```

### 4. Cluster Sampling

Population is divided into clusters, and some clusters are selected.

Example:

```text
Cities → Select some cities → Survey customers
```

# 54. Correlation vs Causation

This is one of the most important concepts for Data Analysts.

### Correlation

Two variables are associated with each other.

Example:

```text
Ice Cream Sales ↑
Temperature ↑
```

Ice cream sales and temperature may be correlated.

### Causation

One variable directly causes a change in another.

Example:

```text
Price Reduction
      ↓
More Purchases
```

A price reduction may cause an increase in purchases if supported by appropriate evidence.

### Important Rule

> **Correlation does not automatically mean causation.**

---

# 55. Example of Correlation Without Direct Causation

Suppose:

```text
Ice Cream Sales ↑
Swimming Pool Visits ↑
```

Both may increase during summer.

A third variable:

```text
Temperature
```

may influence both.

```text
          Temperature
          ↙        ↘
Ice Cream Sales   Pool Visits
```

Therefore, simply observing that two variables move together does not prove that one causes the other.

# 67. Interview Questions from Module 2

### Basic Questions

**Q1. What is data?**

Data is a collection of raw facts, observations, measurements, or values that can be analyzed to generate information and insights.

**Q2. What is the difference between data and information?**

Data is raw and unprocessed, while information is processed data that provides meaning.

**Q3. What is qualitative data?**

Qualitative data represents categories or characteristics, such as city, department, or payment method.

**Q4. What is quantitative data?**

Quantitative data represents numerical quantities, such as age, salary, revenue, or quantity.

**Q5. What is the difference between discrete and continuous data?**

Discrete data is countable, while continuous data is measurable and can take values within a range.

**Q6. What is categorical data?**

Categorical data represents observations belonging to distinct groups or categories.

**Q7. What is an observation?**

An observation is one individual record in a dataset, usually represented by one row.

**Q8. What is a variable?**

A variable is a characteristic or attribute measured for each observation.

**Q9. What is a feature?**

A feature is an input variable used to analyze or predict an outcome, especially in Machine Learning.

**Q10. What is a target variable?**

The target variable is the outcome that a Machine Learning model or analysis aims to predict or explain.

---

### Intermediate Questions

**Q11. What is data granularity?**

Data granularity refers to the level of detail represented by each record in a dataset.

**Q12. Why is granularity important?**

Because incorrect understanding of granularity can lead to incorrect aggregation, double counting, and misleading analysis.

**Q13. What is cardinality?**

Cardinality refers to the number of unique values in a column.

**Q14. What is the difference between nominal and ordinal data?**

Nominal data has categories without a meaningful order, while ordinal data has categories with a meaningful order.

**Q15. What is the difference between interval and ratio data?**

Both have equal intervals, but ratio data has a true zero while interval data does not.

**Q16. What is the difference between population and sample?**

Population represents the complete group being studied, while a sample is a subset of that population.

**Q17. What is an outlier?**

An outlier is an observation that is unusually far from the typical values in a dataset.

**Q18. What is data bias?**

Data bias occurs when the collected or analyzed data systematically misrepresents the population or reality.

**Q19. What is correlation?**

Correlation measures the degree to which two variables are associated or move together.

**Q20. Does correlation imply causation?**

No. Correlation alone does not establish that one variable causes the other.
