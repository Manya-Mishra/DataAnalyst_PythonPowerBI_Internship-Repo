## 📊 Module 3: Statistics for Data Analysis

### 1. What is Statistics?

**Statistics** is the process of collecting, organizing, analyzing, interpreting, and presenting data.

In Data Analysis, statistics helps us answer questions such as:

* What is the average sales?
* How spread out are the values?
* What is the most common category?
* Are sales increasing or decreasing?
* Are two variables related?
* Is a value unusually high or low?
* Can we make conclusions about a larger population from a sample?

#### Simple Flow

```text
Data
 ↓
Collect
 ↓
Organize
 ↓
Summarize
 ↓
Analyze
 ↓
Interpret
 ↓
Make Decisions
```

---

### 2. Types of Statistics

Statistics is broadly divided into two types:

```text
Statistics
│
├── Descriptive Statistics
│
└── Inferential Statistics
```

## Descriptive Statistics

Describes and summarizes the data we already have.

Examples:

```text
Mean
Median
Mode
Range
Variance
Standard Deviation
Percentiles
Quartiles
```

Example:

```text
Sales = [100, 200, 300, 400, 500]

Mean = 300
```

---

## Inferential Statistics

Uses a sample to make conclusions or estimates about a larger population.

Examples:

```text
Hypothesis Testing
Confidence Intervals
Sampling
Statistical Significance
p-value
```

Example:

```text
Survey 1,000 customers
        ↓
Analyze sample
        ↓
Estimate behavior of all customers
```

---

### 3. Descriptive vs Inferential Statistics

| Feature  | Descriptive      | Inferential           |
| -------- | ---------------- | --------------------- |
| Purpose  | Summarize data   | Draw conclusions      |
| Data     | Observed dataset | Usually sample        |
| Examples | Mean, Median, SD | Hypothesis testing    |
| Question | What happened?   | What can we conclude? |

## 6. Mean

The **mean** is the arithmetic average.

### Formula

```text
Mean = Sum of all values / Number of values
```

Mathematically:

```text
x̄ = Σx / n
```

Where:

```text
x̄ = Mean
Σx = Sum of values
n = Number of observations
```

### Example

```text
10, 20, 30, 40, 50
```

Sum:

```text
10 + 20 + 30 + 40 + 50 = 150
```

Number of values:

```text
5
```

Therefore:

```text
Mean = 150 / 5
     = 30
```

### Pandas

```python
df["Salary"].mean()
```

### Python

```python
sum(values) / len(values)
```

---

## 7. Median

The **median** is the middle value after sorting the data.

### Example: Odd Number of Values

```text
10, 20, 30, 40, 50
```

Middle value:

```text
30
```

Therefore:

```text
Median = 30
```

### Example: Even Number of Values

```text
10, 20, 30, 40
```

Two middle values:

```text
20 and 30
```

Median:

```text
(20 + 30) / 2 = 25
```

### Pandas

```python
df["Salary"].median()
```

## 8. Mean vs Median

Consider:

```text
20,000
25,000
30,000
35,000
5,00,000
```

Mean:

```text
(20000 + 25000 + 30000 + 35000 + 500000) / 5
= 122000
```

Median:

```text
30,000
```

The extreme value `5,00,000` greatly increases the mean.

Therefore:

> **Median is generally more resistant to extreme values than mean.**

### When to use?

| Situation                           | Useful measure      |
| ----------------------------------- | ------------------- |
| Data is roughly symmetric           | Mean                |
| Data contains extreme values        | Median              |
| Income/salary data is highly skewed | Median often useful |
| Need mathematical calculations      | Mean                |

## 13. Variance

Variance measures how far values are spread from the mean.

### Basic idea

```text
Small variance
→ Values are close to mean

Large variance
→ Values are more spread out
```

### Population Variance

```text
σ² = Σ(x - μ)² / N
```

### Sample Variance

```text
s² = Σ(x - x̄)² / (n - 1)
```

Where:

```text
μ = Population mean
x̄ = Sample mean
N = Population size
n = Sample size
```

---

## 14. Variance Example

Consider:

```text
2, 4, 6
```

Mean:

```text
(2 + 4 + 6) / 3 = 4
```

Calculate deviations:

```text
2 - 4 = -2
4 - 4 =  0
6 - 4 = +2
```

Square them:

```text
4, 0, 4
```

Sum:

```text
8
```

Population variance:

```text
8 / 3
= 2.67
```

### Pandas

Sample variance:

```python
df["Salary"].var()
```

Population variance:

```python
df["Salary"].var(ddof=0)
```

## 15. Standard Deviation

Standard deviation measures how far values typically spread from the mean.

It is the square root of variance.

### Formula

```text
Standard Deviation = √Variance
```

### Example

If:

```text
Variance = 25
```

Then:

```text
Standard Deviation = √25
                   = 5
```

### Pandas

```python
df["Salary"].std()
```

Population standard deviation:

```python
df["Salary"].std(ddof=0)
```

---

## 16. Why Standard Deviation is Important

Suppose two classes have the same average score:

```text
Class A:
70, 71, 69, 70, 70

Class B:
40, 60, 70, 80, 100
```

Both may have:

```text
Mean = 70
```

But Class B has much greater spread.

Therefore:

> Mean alone does not tell us the complete story.

We should also examine:

```text
Standard Deviation
```

```text
Q1 → 25th percentile
Q2 → 50th percentile
Q3 → 75th percentile
```

### Q2

### Q3

---

## 20. Interquartile Range — IQR

IQR measures the spread of the middle 50% of the data.

### Formula

```text
IQR = Q3 - Q1
```

Example:

```text
Q1 = 20
Q3 = 50
```

Therefore:

```text
IQR = 50 - 20
    = 30
```

### Pandas

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1
```

---

## 21. IQR and Outlier Detection

A common rule uses:

```text
Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR
```

Values outside these limits may be considered outliers.

### Python

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Salary"] < lower) |
    (df["Salary"] > upper)
]
```

### Important

The `1.5 × IQR` rule is a common rule for identifying potential outliers. It does not automatically mean every flagged value is an error.

## 27. Normal Distribution

A normal distribution is a symmetric, bell-shaped distribution.

It approximately looks like:

```text
                 *
              *     *
            *         *
          *             *
        *                 *
______*_____________________*______
```

In a perfectly normal distribution:

```text
Mean ≈ Median ≈ Mode
```

---

## 28. Properties of Normal Distribution

Important properties:

```text
Symmetric
Bell-shaped
Mean = Median = Mode
Total area under curve = 1
```

The normal distribution is important in statistics because many statistical methods are based on it or approximate it under certain conditions.

---

## 29. Empirical Rule — 68-95-99.7 Rule

For approximately normally distributed data:

```text
~68% → within 1 standard deviation

~95% → within 2 standard deviations

~99.7% → within 3 standard deviations
```

Example:

Suppose:

```text
Mean = 100
SD = 10
```

Approximately:

```text
68% → 90 to 110
95% → 80 to 120
99.7% → 70 to 130
```

This rule applies to distributions that are approximately normal.

## 31. Right-Skewed Distribution

## 35. Z-Score

A **Z-score** tells us how many standard deviations an observation is away from the mean.

### Formula

```text
Z = (X - μ) / σ
```

Where:

```text
X = Observation
μ = Mean
σ = Standard deviation
```

### Example

Suppose:

```text
Mean = 50
SD = 10
X = 70
```

Then:

```text
Z = (70 - 50) / 10
  = 2
```

So:

```text
70 is 2 standard deviations above the mean.
```

---

## 36. Interpretation of Z-Score

```text
Z = 0
→ Exactly at mean

Z = +1
→ 1 SD above mean

Z = -1
→ 1 SD below mean

Z = +2
→ 2 SD above mean

Z = -2
→ 2 SD below mean
```

Large absolute Z-scores can indicate unusual observations, but the appropriate threshold depends on the distribution and context.

## 45. Probability Basics

Probability measures how likely an event is to occur.

Its value lies between:

```text
0 and 1
```

or:

```text
0% and 100%
```

#### Formula

```text
P(Event) =
Number of favorable outcomes /
Total possible outcomes
```

---

## 46. Probability Example

A fair coin has:

```text
Head
Tail
```

Probability of Head:

```text
P(Head) = 1 / 2
        = 0.5
        = 50%
```

Probability of Tail:

```text
P(Tail) = 50%
```

## 47. Basic Probability Rules

#### Rule 1: Probability Range

```text
0 ≤ P(A) ≤ 1
```

#### Rule 2: Impossible Event

```text
P(A) = 0
```

#### Rule 3: Certain Event

```text
P(A) = 1
```

---

## 48. Complement

The probability that an event does **not** happen is:

```text
P(A') = 1 - P(A)
```

Example:

If:

```text
P(Customer Churns) = 0.25
```

Then:

```text
P(Customer Does Not Churn)
= 1 - 0.25
= 0.75
```

or:

```text
75%
```

```text
P(A | B)
```

Read as:

#### Example

Suppose:

```text
P(Customer Churns | Month-to-Month Contract)
```

```text
Customer Analytics
Risk Analysis
Fraud Detection
Marketing
```

## 50. Expected Value

Expected value represents the long-run average outcome of a random process.

#### Formula

```text
E(X) = Σ [x × P(x)]
```

Example:

Suppose:

```text
Outcome = ₹100
Probability = 0.5

Outcome = ₹200
Probability = 0.5
```

Expected value:

```text
E(X)
= (100 × 0.5) + (200 × 0.5)
= 50 + 100
= ₹150
```

Sampling bias occurs when the sample does not properly represent the population.

## 55. Sampling Distribution

A **sampling distribution** is the distribution of a statistic obtained from repeated samples from the same population.

For example:

```text
Population
    ↓
Sample 1 → Mean
Sample 2 → Mean
Sample 3 → Mean
Sample 4 → Mean
...
    ↓
Distribution of Sample Means
```

This concept becomes important for:

```text
Confidence Intervals
Hypothesis Testing
Central Limit Theorem
```

---

## 56. Central Limit Theorem — CLT

The **Central Limit Theorem** is a fundamental statistical concept.

In simplified terms:

> For sufficiently large random samples, the sampling distribution of the sample mean tends to become approximately normal under common conditions, even when the original population is not normally distributed.

#### Concept

```text
Population
    ↓
Random Samples
    ↓
Calculate Mean of Each Sample
    ↓
Distribution of Sample Means
    ↓
Approximately Normal
```

The exact conditions and required sample size depend on the underlying population and sampling process.

---

## 57. Standard Error

Standard error measures the variability of a sample statistic across repeated samples.

For the sample mean:

```text
SE = s / √n
```

Where:

```text
s = Sample standard deviation
n = Sample size
```

As sample size increases:

```text
n ↑
↓
SE ↓
```

So larger samples generally provide more precise estimates of the population mean, assuming appropriate sampling.

---

## 58. Confidence Interval

A confidence interval provides a range of plausible values for a population parameter based on sample data.

Example:

```text
Average customer spending
= ₹500

95% Confidence Interval:
₹480 – ₹520
```

The interval communicates uncertainty around the estimate.

### Important Interpretation

A 95% confidence interval is produced by a procedure that, under repeated sampling and its assumptions, captures the true parameter about 95% of the time.

Avoid interpreting it simply as:

> "There is a 95% probability that the fixed parameter is inside this particular interval."

---

## 59. Confidence Level

Common confidence levels:

Higher confidence generally produces a wider interval, all else being equal.

```text
Confidence ↑
     ↓
Interval Width ↑
```

## 60. Hypothesis Testing

Hypothesis testing helps determine whether sample evidence provides enough evidence against a specified assumption about a population.

Basic structure:

```text
Problem
  ↓
Null Hypothesis
  ↓
Alternative Hypothesis
  ↓
Collect Data
  ↓
Statistical Test
  ↓
p-value / Test Statistic
  ↓
Decision
```

---

# 61. Null Hypothesis

The **null hypothesis**, represented by `H₀`, usually states that there is no effect, difference, or association according to the tested claim.

Example:

```text
H₀:
New website design does not change average purchase value.
```

---

# 62. Alternative Hypothesis

The **alternative hypothesis**, represented by `H₁` or `Hₐ`, represents the competing claim.

Example:

```text
H₁:
New website design changes average purchase value.
```

---

# 63. p-value

A **p-value** is the probability, assuming the null hypothesis is true, of obtaining a result at least as extreme as the observed result according to the chosen test.

A small p-value can provide evidence against the null hypothesis.

A common threshold is:

```text
α = 0.05
```

But the threshold should be chosen based on the analysis context rather than blindly applying 0.05.

### Simplified decision rule

```text
p-value < α
→ Reject H₀

p-value ≥ α
→ Fail to reject H₀
```

### Important

Do **not** say:

```text
p-value > 0.05
→ H₀ is proven true
```

Instead say:

```text
There is insufficient evidence to reject H₀.
```

---

# 64. Statistical Significance

A result is often called **statistically significant** when the test result provides sufficient evidence against the null hypothesis at a pre-specified significance level.

Example:

```text
α = 0.05
p-value = 0.02
```

Since:

```text
0.02 < 0.05
```

the result is statistically significant at the 5% level.

### Important

Statistical significance does **not** automatically mean:

```text
Large business impact
Practical importance
Causal relationship
```

Always consider effect size and business context.

---

# 65. Type I and Type II Errors

Hypothesis testing can result in two major types of errors.

### Type I Error

Rejecting a true null hypothesis.

```text
False Positive
```

Example:

Concluding that a new feature has an effect when it actually does not.

### Type II Error

Failing to reject a false null hypothesis.

```text
False Negative
```

Example:

Concluding that there is not enough evidence of an effect when a real effect exists.

---

# 66. Type I vs Type II

| Error   | Meaning                 | Common term    |
| ------- | ----------------------- | -------------- |
| Type I  | Reject true H₀          | False Positive |
| Type II | Fail to reject false H₀ | False Negative |

Memory:

```text
Type I  → False Positive
Type II → False Negative
```

# 67. A/B Testing

A/B testing is a controlled experiment used to compare two versions.

Example:

```text
Version A → Existing Website
Version B → New Website
```

Users are randomly assigned to groups.

Then compare:

```text
Conversion Rate
Revenue
Click-through Rate
Retention
```

### Flow

```text
Users
  ↓
Random Assignment
  ↓
 ┌──────────────┐
 ↓              ↓
Group A       Group B
Old Version   New Version
 ↓              ↓
Metric A      Metric B
 └──────┬───────┘
        ↓
Statistical Analysis
        ↓
Conclusion
```

---

# 68. Example of A/B Testing

Suppose:

```text
Version A:
1000 visitors
50 purchases

Conversion Rate = 5%

Version B:
1000 visitors
70 purchases

Conversion Rate = 7%
```

Observed difference:

```text
7% - 5% = 2 percentage points
```

But we should not conclude that Version B caused a real improvement solely from these observed rates.

We should consider:

```text
Statistical test
Confidence interval
Sample size
Randomization
Effect size
Business significance
```

---

# 69. Effect Size

Effect size describes the **magnitude of a difference or relationship**.

This is important because:

```text
Statistically significant
```

does not necessarily mean:

```text
Practically important
```

Example:

A new feature increases conversion from:

```text
10.00% → 10.01%
```

With a huge sample, this difference might be statistically significant but have limited practical impact.

Therefore, analyze:

```text
Statistical significance
+
Effect size
+
Business impact
```

# 82. Important Formulas

## Mean

```text
x̄ = Σx / n
```

## Range

```text
Range = Max - Min
```

## Population Variance

```text
σ² = Σ(x - μ)² / N
```

## Sample Variance

```text
s² = Σ(x - x̄)² / (n - 1)
```

## Standard Deviation

```text
SD = √Variance
```

## IQR

```text
IQR = Q3 - Q1
```

## Outlier Bounds

```text
Lower = Q1 - 1.5 × IQR

Upper = Q3 + 1.5 × IQR
```

## Z-score

```text
Z = (X - μ) / σ
```

## Complement

```text
P(A') = 1 - P(A)
```

## Expected Value

```text
E(X) = Σ[x × P(x)]
```

## Standard Error of Mean

```text
SE = s / √n
```

# 84. Interview Questions

### Q1. What is the difference between mean and median?

**Mean** is the arithmetic average, while **median** is the middle value after sorting the observations.

---

### Q2. When would you prefer median over mean?

When the distribution is highly skewed or contains influential extreme values, median can provide a more representative measure of the typical observation.

---

### Q3. What is standard deviation?

Standard deviation measures the typical spread of observations around the mean.

---

### Q4. What is the difference between variance and standard deviation?

Variance is based on squared deviations from the mean, while standard deviation is the square root of variance and therefore has the same units as the original variable.

---

### Q5. What is IQR?

IQR is:

```text
Q3 - Q1
```

and represents the spread of the middle 50% of observations.

---

### Q6. How do you detect outliers?

Common approaches include:

```text
IQR method
Z-score
Box plot
Domain/business rules
```

---

### Q7. What is correlation?

Correlation measures the direction and strength of a linear relationship between two numerical variables.

---

### Q8. What is the range of Pearson correlation?

```text
-1 to +1
```

---

### Q9. Does correlation imply causation?

No. Correlation indicates association, not necessarily a causal relationship.

---

### Q10. What is a p-value?

A p-value measures how unusual the observed result, or something more extreme, would be if the null hypothesis were true, according to the statistical test being used.

---

### Q11. What is a confidence interval?

A confidence interval is an interval estimate produced by a statistical procedure to quantify uncertainty around a population parameter.

---

### Q12. What is a Type I error?

Rejecting a true null hypothesis.

```text
False Positive
```

---

### Q13. What is a Type II error?

Failing to reject a false null hypothesis.

```text
False Negative
```

---

### Q14. What is A/B testing?

A/B testing is a controlled experiment in which different groups are exposed to different versions and their outcomes are compared using predefined metrics and statistical analysis.

---

### Q15. Why is sample size important?

Larger appropriate samples generally reduce sampling variability and improve the precision of estimates, although sample quality and study design remain important.
