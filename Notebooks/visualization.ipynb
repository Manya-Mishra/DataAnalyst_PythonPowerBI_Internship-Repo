# 1. What is Data Visualization?

## Definition

Data Visualization is the graphical representation of data using charts, graphs, maps, and dashboards to communicate information clearly and efficiently.

It helps identify:
- Trends
- Patterns
- Relationships
- Outliers
- Business Insights

## Example

### Raw Data

| Month | Sales |
|---------|-------|
| Jan | 50 |
| Feb | 60 |
| Mar | 75 |
| Apr | 90 |

Looking at the table, it's difficult to identify the trend.

Using a **Line Chart**, we can immediately see that sales are increasing over time.

# 6. When to Use Which Chart

### <U>Choosing the Right Chart in Data Analysis</U>

One of the most important skills of a Data Analyst is selecting the appropriate chart. Choosing the wrong chart can make insights difficult to understand or even misleading.


#### Quick Decision Guide

| If you want to... | Use This Chart |
|-------------------|---------------|
| Compare categories | Bar Chart |
| Show trends over time | Line Chart |
| Show percentage/share | Pie Chart |
| Show relationship between two numerical variables | Scatter Plot |
| Show data distribution | Histogram |
| Detect outliers | Box Plot |
| Compare distributions | Box Plot / Violin Plot |
| Compare parts of a whole over time | Stacked Bar Chart |
| Show correlation between many variables | Heatmap |
| Compare values across two categories | Grouped Bar Chart |
| Show cumulative total | Area Chart |
| Display exact values | Table |


# 1. <B>Bar Chart</B>

## Purpose

Compare values across different categories.

## Best For

- Sales by Product
- Revenue by City
- Students in Each Department
- Number of Employees per Department

## Example Dataset

| Product | Sales |
|----------|------:|
| Laptop | 500 |
| Mobile | 800 |
| Tablet | 300 |
| Printer | 200 |

### Best Chart

**Bar Chart**

Reason:
Each product is a different category.

### Pandas

```python
df.plot(
    x="Product",
    y="Sales",
    kind="bar"
)
```

# 2. <B>Line Chart</B>

## Purpose

Show trends over time.

## Best For

- Monthly Sales
- Daily Temperature
- Stock Prices
- AQI Trends

## Example

| Month | Revenue |
|-------|---------:|
| Jan | 50 |
| Feb | 60 |
| Mar | 75 |
| Apr | 80 |

### Best Chart

**Line Chart**

Reason:
Months follow chronological order.

### Pandas

```python
df.plot(
    x="Month",
    y="Revenue",
    kind="line"
)
```

# 3. <B>Pie Chart</B>

## Purpose

Show percentage contribution of each category to the whole.

## Best For

- Market Share
- Budget Allocation
- Revenue Contribution
- Gender Distribution

## Example

| Category | Revenue |
|----------|---------:|
| Electronics | 45 |
| Clothing | 30 |
| Grocery | 25 |

### Best Chart

**Pie Chart**

Reason:
The categories together make up 100%.

### Pandas

```python
df.plot(
    y="Revenue",
    kind="pie",
    labels=df["Category"],
    autopct="%1.1f%%"
)
```

### Avoid Pie Charts When

- More than 5–6 categories
- Comparing exact values
- Showing trends over time

# 7. <B>Heatmap</B>

## Purpose

Visualize correlation between variables.

## Best For

- Feature Correlation
- EDA
- Machine Learning

## Example

| | Age | Salary | Experience |
|---|----:|-------:|-----------:|
| Age | 1.0 | 0.72 | 0.84 |
| Salary | 0.72 | 1.0 | 0.91 |
| Experience | 0.84 | 0.91 | 1.0 |

### Best Chart

**Heatmap**

Reason:
Shows strong and weak correlations using colors.

### Seaborn

```python
import seaborn as sns

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True
)
```

# 7. Comparison of Similar Charts

## <B>Bar Chart vs Histogram</B>

| Bar Chart | Histogram |
|------------|-----------|
| Categorical Data | Continuous Data |
| Gaps between bars | No gaps |
| Compare categories | Show distribution |


## <B>Pie Chart vs Donut Chart</B>

| Pie | Donut |
|------|--------|
| Simpler | Better for dashboards |
| Hard with many categories | Can display total in center |


## <B>Scatter Plot vs Bubble Chart</B>

| Scatter | Bubble |
|----------|--------|
| Two variables | Three variables |
| Equal point size | Bubble size represents third variable |


## <B>Area Chart vs Line Chart</B>

| Line | Area |
|------|------|
| Focus on trend | Focus on trend + magnitude |

# 18. Mini Case Study

### Business Problem

An online retail company wants to improve sales.

### Data Collected

- Monthly Revenue
- Product Category
- Customer Region
- Profit
- Order Count

### Visualizations Used

- KPI Cards → Total Revenue, Total Profit
- Line Chart → Monthly Revenue Trend
- Bar Chart → Sales by Category
- Pie Chart → Region-wise Revenue Share
- Heatmap → Correlation Between Revenue and Profit
- Box Plot → Detect Outliers in Order Value

### Insights

- Sales peak during festive months.
- Electronics generate the highest revenue.
- West region contributes the largest share.
- A few orders have unusually high values (outliers).

### Business Recommendation

Increase inventory before festive seasons, focus marketing on high-performing categories, and investigate unusually large orders.
